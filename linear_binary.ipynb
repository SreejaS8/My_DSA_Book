```python
def locate_card(cards, query):
    position = 0
    while True:
        if cards[position] == query:
            return position
        position += 1
        if position == len(cards):
            return -1

#card = [13, 11, 10, 7, 4, 3, 1, 0]
#query = 7
#locate_card(card, query)

test = {
    'input':{
        'cards': [13, 11, 10, 7, 4, 3, 1, 0],
        'query': 7
    },
    'output': 3
}

result = locate_card(**test['input'])
result
```
